# Exploratory Data Analysis: Rick and Morty Dialogue — Speaker Prediction

**Problem statement.** This notebook explores a transcript-derived dataset of spoken lines from the animated
series *Rick and Morty*. Each row is one line of dialogue tagged with the episode it belongs to and the
character who said it. The long-term goal of this project is to build a model that predicts **who said a
given line of dialogue**. Before any modeling is attempted, this notebook performs a full exploratory data
analysis: understanding how the dataset is built, what each row represents, what data-quality problems exist,
which patterns are relevant to the prediction task, and which decisions will need to be validated later during
modeling.


## 1. Context and Problem Definition

**Domain.** *Rick and Morty* is an animated sitcom. The dataset used here is a scraped transcript: every
spoken (or narrated) line across a subset of episodes, paired with the episode number and the speaker tag
assigned by whoever produced the transcript.

**Motivating question.** Given **only the text of a line of dialogue**, *who is the character most likely to
have said it?* This is motivated by the idea that characters in a well-written show have a distinguishable voice (vocabulary, syntax, verbosity, catchphrases)
that a model might learn to recognize.

**Expected usefulness.** A working speaker-prediction model would be a concrete demonstration that character
voice is statistically learnable from text alone. It also has a practical byproduct: it can help clean/verify
speaker labels in similar scraped transcripts, whose labels are noisy (this EDA finds formatting variants,
HTML-wrapped names and labels that are not characters at all).

**Unit of observation.** After loading and lightly cleaning the data, **one row = one attributed line of
dialogue**: a contiguous block of text assigned to a single speaker within a single episode. Some rows mix
narration/stage directions with the actual spoken text (addressed in Section 5 and Section 9); the *spoken
content* of the row, not the narration, is what should ultimately feed a speaker-prediction model.

**Target variable.** The target is `speaker` (character name who delivered the line, cleaned into
`speaker_clean`; consolidated framings such as `speaker_group` are evaluated as experimental alternatives in
Section 9.1). Its meaning is literal: the name
of the character credited with the line.

**Prediction-time information set.** The premise fixes what a model may use when it makes a prediction:

| Role | What it is | Available when predicting? |
|------|------------|----------------------------|
| **Input** | The text of a single dialogue line (and anything derived only from that text) | Yes |
| **Target** | The character who said the line (`speaker`) | No (it is what we predict) |
| **Not an input** | `episode no.`, the line's position inside the episode, neighbouring lines, episode-level statistics | No |

`episode no.` is therefore used in this notebook **only** for exploration (coverage, data quality) and to design a
leakage-safe validation split later; it is **never** proposed as a predictor for the main model. A model that
also receives episode or conversational context would be a *different* problem (a separate "contextual"
experiment) and is out of scope here.

**Type of ML problem: multiclass classification.** Given a dialogue line, predict which one of several
possible speakers said it. It is **not** binary (many possible characters) and **not** multilabel (each line
has exactly one speaker). As Section 7 will show, the raw target has a very large number of classes with a
severe long tail (many characters appear only once or twice), which has direct consequences for how the
classification problem should be framed for modeling. This is a question the EDA raises but does not resolve,
since resolving it requires decisions (e.g. class consolidation thresholds) that belong in the modeling stage.


### 1.1 Guiding Questions

This EDA is organized around the following concrete questions, adapted to this dataset (not copied from the
Titanic reference):

1. What variables does the dataset contain, and what is each one's analytical role (identifier, categorical,
   free text, sequential index)? Which of them are actually available when only a dialogue line is given?
2. Are there missing values, duplicated rows, or malformed entries (scrape artifacts), and how severe are they?
3. How consistent is the `speaker` field? Does the same character appear under multiple spellings/formats?
4. How is the target variable (`speaker`) distributed? Is it balanced enough for direct multiclass modeling,
   or does it need consolidation, and what majority-class baseline does each framing imply?
5. Do lines mix narration/stage directions with actual spoken text, and how much of it can be separated from
   the spoken words that should feed the model?
6. Does dialogue length, punctuation, vocabulary (single words and two-word expressions), or interjection use
   differ across speakers?
7. How often do lines mention other characters by name, and how much of the apparent lexical signal is the
   character's own **voice** versus **conversational context** (who they usually talk to)?
8. How does speaker prominence vary across episodes, and what does that imply for how a validation split should
   be designed (rather than for which features to use)?
9. Which text-derived variables are worth carrying into modeling, and which candidate variables must be excluded
   because they would not exist at prediction time?

### 3.1 Temporal / Sequential Structure

`episode no.` gives the dataset a natural order: dialogue lines within an episode are already in speaking
order (row order == line order, since this is a transcript), and episodes themselves have a natural numeric
order. Any analysis that looks at "how things change over the course of the show" must respect this order:
rows must **not** be shuffled before such analysis.

Per the prediction-time information set (Section 1), the episode number and the line's position are **not**
model inputs. The sequential structure still matters for two reasons: it is needed to describe the data
correctly (coverage, gaps), and it informs how a validation split should be built. Lines from the same episode
share plot-specific vocabulary (guest names, objects, locations), so a random row-level split would let the
model see an episode's vocabulary in training and be evaluated on lines from that same episode, an optimistic
form of leakage. Grouping the split by episode is a hypothesis to evaluate during modeling (Section 11).

Before trusting `episode no.` as a reliable sequential axis, its coverage needs to be checked for gaps.


```python
ep_min, ep_max = df["episode no."].min(), df["episode no."].max()
present_eps = set(df["episode no."].unique())
full_range = set(range(ep_min, ep_max + 1))
missing_eps = sorted(full_range - present_eps)

rows_per_ep = df.groupby("episode no.").size().sort_index()

print(f"Episode numbers span {ep_min}-{ep_max} ({len(full_range)} possible values).")
print(f"{len(present_eps)} distinct episode numbers are actually present.")
print(f"Missing episode numbers: {missing_eps}")
print()
print("Rows per episode, summary statistics:")
print(rows_per_ep.describe())
```

```python
fig, ax = plt.subplots(figsize=(11, 4.5))
rows_per_ep.plot(kind="bar", ax=ax, color="#4C72B0", width=0.8)
ax.set_title("Number of Dialogue Rows per Episode")
ax.set_xlabel("Episode number")
ax.set_ylabel("Row count")
ax.tick_params(axis="x", labelrotation=90)
plt.tight_layout()
plt.show()
```

**Interpretation.** Episode numbers run from 1 to 51, but only 39 distinct values are present: 12 episode
numbers (`31, 32, 33, 34, 36, 38, 40, 41, 44, 45, 46, 50`) have **no rows at all**, i.e. those episodes were
never scraped rather than being empty. Row counts per episode are also highly uneven (from as few as ~14 rows
to over 800), which does not look like a clean per-season structure; it looks like a scrape that thinned out
for later episodes, plausibly stopping partway through some of them. This matters for two reasons: (1) any
"trend across the show" analysis is really a trend across an *incomplete, unevenly sampled* subset of episodes,
and (2) episodes with very few rows should not be trusted for per-episode conclusions (small-sample caution,
revisited in Section 8). This is recorded as a limitation in Section 11.

## 5. Data Quality Diagnostics

This section diagnoses problems without fixing them silently: every issue found here is quantified first,
interpreted second, and only turned into an actual fix when it is a pure formatting/normalization issue with
clear before/after evidence (never by dropping or imputing rows without justification).


```python
missing = df.isna().sum()
missing_pct = (df.isna().mean() * 100).round(2)
pd.DataFrame({"n_missing": missing, "pct_missing": missing_pct})
```

```python
missing_speaker_idx = df.index[df["speaker"].isna()]
lo, hi = max(missing_speaker_idx.min() - 1, 0), min(missing_speaker_idx.max() + 1, len(df) - 1)
df.loc[lo:hi].assign(is_missing=lambda d: d.index.isin(missing_speaker_idx))
```

**Interpretation: missing speakers.** Only 3 rows (0.03% of the data) are missing a `speaker` value, and
all 3 sit inside the same episode, next to rows that do have a speaker, consistent with a scrape gap in one
scene rather than a systemic problem. Because there are so few, and because guessing the speaker from
neighboring context would be exactly the kind of unjustified imputation the assignment warns against, these
rows are **kept with `speaker` as missing** and simply excluded from any analysis that groups by speaker.


```python
dup_mask = df.duplicated(subset=["episode no.", "speaker", "dialogue"], keep=False)
dup_count = df.duplicated(subset=["episode no.", "speaker", "dialogue"], keep="first").sum()
print(f"{dup_count:,} rows are exact duplicates of another row (same episode, speaker and dialogue text).")

HTML_TAG_RE = r"<[a-zA-Z/][^>]*>"
html_in_dialogue = df["dialogue"].str.contains(HTML_TAG_RE, na=False)
html_in_speaker = df["speaker"].str.contains(HTML_TAG_RE, na=False)
print(f"{html_in_dialogue.sum():,} dialogue values contain raw HTML tags; by episode: "
      f"{df.loc[html_in_dialogue, 'episode no.'].value_counts().to_dict()}")
print(f"{html_in_speaker.sum():,} speaker values contain raw HTML tags, e.g. {df.loc[html_in_speaker, 'speaker'].iloc[0]!r}")

junk_mask = df["dialogue"].str.strip().isin(["", "["])
print(f"{junk_mask.sum():,} rows look like scrape junk (empty/whitespace-only or a stray '[').")
df.loc[junk_mask, ["episode no.", "speaker", "dialogue"]].head(8)
```

**Interpretation: duplicates.** Several hundred rows are exact duplicates of another row on
episode+speaker+dialogue. A sample of the duplicated content shows two different situations: (a) genuinely
repeated short lines (e.g. a character saying "What?" twice, which can legitimately happen twice in a
transcript) and (b) clear scrape junk: empty strings or a lone `"["`. These are
**not removed automatically**: junk rows are a real quality problem to flag for the modeling stage (they carry
no usable text and would need dropping before vectorization), while short genuine repeats are valid
observations that happen to collide in this particular projection. Blindly de-duplicating would silently
delete legitimate repeated lines along with the junk.

**Interpretation: raw HTML markup.** A separate scrape artifact is literal HTML left inside the text. In the
dialogue it is concentrated in one episode (episode 12, where almost every line is wrapped in
`<span style="font-weight: normal">`), with a few `<br>`/`<i>` tags elsewhere. In the speaker column it wraps
some names (`<span style="font-style: normal">Beth:</span>`), which would silently create a second "Beth"
class. Both are pure markup with no meaning: they are stripped in the speaker normalization below and in the
dialogue cleaning of Section 9.2. Left in place, words like *span* and *style* would enter the vocabulary as if
a character had said them, and only in the episode that happens to contain the markup.


```python
dialogue_len_chars = df["dialogue"].str.len()
longest_idx = dialogue_len_chars.idxmax()
shortest_nonempty_idx = dialogue_len_chars[dialogue_len_chars > 0].idxmin()

print("Longest dialogue entry (character count = {:,}):".format(dialogue_len_chars.max()))
print(df.loc[longest_idx, "dialogue"][:400], "...\n")

print("Shortest non-empty dialogue entry:")
print(repr(df.loc[shortest_nonempty_idx, "dialogue"]))
```

**Interpretation: length extremes.** The longest entry (~1,800 characters) is a genuine long monologue
with several sentences run together, not a merge artifact, just a valid observation with a heavy tail.
Extremely short entries (single characters/words like `"Oh."`) are also legitimate: real dialogue lines are
often that short. No rows are removed on length grounds alone; the distribution itself is analyzed in Section 6.


```python
def text_markers(s: pd.Series) -> pd.DataFrame:
    r"""Boolean flags per row. The first three are layout artifacts; the rest are signs of real narration.

    Whitespace is written as [\s\xa0] so results do not depend on the pandas regex engine
    (with pyarrow installed, \s is ASCII-only and would miss non-breaking spaces).
    """
    return pd.DataFrame({
        "leading_colon (layout)": s.str.match(r"^[\s\xa0]*:"),
        "newline/indent/padding (layout)": s.str.contains(r"\n|^[\s\xa0]|[\s\xa0]$|[\s\xa0]{2,}"),
        "html_tags (layout)": s.str.contains(HTML_TAG_RE),
        "[...], (...) or *...* (narration)": s.str.contains(r"[\[\]()*]"),
        "lowercase action before speech (narration)": s.str.startswith("\n") & s.str.match(r"^[\s\xa0]*[a-z]"),
    })


markers_raw = text_markers(df["dialogue"])
is_layout_artifact = markers_raw.iloc[:, :3].any(axis=1)
has_narration_marker = markers_raw.iloc[:, 3:].any(axis=1)

marker_summary = markers_raw.mean().mul(100).round(1).rename("pct_of_rows").to_frame()
marker_summary.loc["ANY layout artifact"] = round(is_layout_artifact.mean() * 100, 1)
marker_summary.loc["ANY narration marker"] = round(has_narration_marker.mean() * 100, 1)
marker_summary
```

**Interpretation: layout artifacts vs. narration mixed into dialogue.** A first, broad check (any leading
whitespace, newline or bracket) flags well over half of the rows, but that number mixes two very different
problems and overstates the second one:

- **Layout artifacts (about 93% of rows).** About 41% of lines start with a stray `:` (a leftover from
  splitting `"Speaker: line"`), most lines carry newlines, indentation or padding from the HTML pages, and a few hundred still contain raw
  HTML tags. These
  change no words; they are fixed by normalization and carry no information about the speaker.
- **Real narration (about 14% of rows).** Bracketed or parenthesized actions and sound cues (`[Grunts]`,
  `(Rips off his shirt.)`, `[bleep]`), `*sighs*`-style actions, and, in a few episodes, a lowercase action
  description written *before* the spoken words (e.g. *"stumbles in drunkenly, and turns on the lights. Morty! You
  gotta come on."*).

Only the second group changes what the "dialogue" input means: a speaker classifier should learn from what a
character **says**, not from a transcriber's description of what they do. Section 8 checks whether narration
is spread evenly, and Section 9.2 builds and evaluates a cleaning step for both groups.

```python
raw_speaker_nunique = df["speaker"].nunique()
trailing_colon = df["speaker"].str.match(r".*:[\s\xa0]*$", na=False)
has_nbsp_or_ws = df["speaker"].str.contains(r"\xa0", regex=True, na=False) | (
    df["speaker"].str.strip() != df["speaker"]
)

print(f"Raw unique speaker values: {raw_speaker_nunique}")
print(f"Rows where speaker has a trailing colon: {trailing_colon.sum():,} ({trailing_colon.mean()*100:.1f}%)")
print(f"Rows with stray whitespace/non-breaking-space in speaker: {has_nbsp_or_ws.sum():,}")

_key = (
    df["speaker"].str.replace(HTML_TAG_RE, "", regex=True).str.replace("\xa0", " ", regex=False).str.strip().str.rstrip(":").str.strip().str.lower()
)
example_groups = (
    df.assign(_key=_key)
    .dropna(subset=["speaker"])
    .groupby("_key")["speaker"]
    .agg(lambda s: sorted(set(s)))
)
example_groups = example_groups[example_groups.apply(len) > 1]
print(f"\n{len(example_groups)} normalized names absorb more than one raw spelling. Examples:")
example_groups.head(6)
```

**Interpretation: speaker formatting.** ~976 distinct raw `speaker` values, but a large share of that
count is inflated by formatting noise, not real characters: about a fifth of all rows have a trailing colon on
the speaker name (a parsing artifact: `"Rick:"` and `"Rick"` are the same character), a further set carry
stray whitespace or non-breaking-space characters, and a few are wrapped in HTML tags. This is a pure text-normalization problem (not a semantic
ambiguity), so, unlike the duplicate/missing-value cases above, it is fixed directly below, with full
before/after evidence, rather than deferred.


```python
_key = df["speaker"].str.replace(HTML_TAG_RE, "", regex=True).str.replace("\xa0", " ", regex=False).str.strip().str.rstrip(":").str.strip().str.lower()
_display = df["speaker"].str.replace(HTML_TAG_RE, "", regex=True).str.replace("\xa0", " ", regex=False).str.strip().str.rstrip(":").str.strip()

canonical_map = (
    df.assign(_key=_key, _display=_display)
    .dropna(subset=["_key"])
    .groupby("_key")["_display"]
    .agg(lambda s: s.value_counts().idxmax())
)
df["speaker_clean"] = _key.map(canonical_map)

print(f"Unique speakers before cleaning: {df['speaker'].nunique()}")
print(f"Unique speakers after cleaning:  {df['speaker_clean'].nunique()}")
df[["speaker", "speaker_clean"]].drop_duplicates(subset=["speaker"]).query("speaker != speaker_clean").head(8)
```

Normalizing whitespace/colon formatting and merging case-only variants (e.g. `"RIck"` -> `"Rick"`)
brings the unique-speaker count down noticeably. The original `speaker` column is kept untouched for audit
purposes; **`speaker_clean` is used from here on** for anything that groups or counts by character.

### 9.1 `speaker_group`: an experimental, capped version of the target

**Definition.** `speaker_clean` collapsed to the top-N most frequent speakers (by line count), with everyone
else mapped to `"Other"`. N=15 is used below purely to illustrate the effect. This is **one candidate framing
of the target, not the target**: the right framing depends on the project goal and must be compared during
modeling.

**Intuition.** Section 7 showed that a literal ~840-class target is impractical. Capping the class count trades
away the ability to name every minor character in exchange for a target that a classifier can actually learn.

```python
TOP_N = 15
top_n_names = target.value_counts().head(TOP_N).index
df["speaker_group"] = df["speaker_clean"].where(df["speaker_clean"].isin(top_n_names), other="Other")
df.loc[df["speaker_clean"].isna(), "speaker_group"] = np.nan

print(f"Classes before consolidation: {df['speaker_clean'].nunique()}")
print(f"Classes after consolidation (top {TOP_N} + Other): {df['speaker_group'].nunique()}")
df["speaker_group"].value_counts(normalize=True).mul(100).round(1).head(TOP_N + 1)
```

**Evidence.** With N=15, the target drops from ~840 classes to 16 and most rows keep their real character
label. But the consolidation has a cost that is easy to miss: `"Other"` becomes the **largest class (about
34% of lines)**, larger than Rick. That changes the reference point: the majority-class baseline for this
framing is about 34% (always predict "Other"), not the 21.7% of `speaker_clean`. Model accuracies on different
framings are therefore **not comparable** unless each is reported against its own baseline.

**Limitations.** `"Other"` is not a character: it lumps hundreds of unrelated voices together, so its internal
variability is huge and a model can learn it as "does not sound like any main character" rather than as a
speaker. The comparison below lays out the alternative framings with their own baselines.

```python
labelled = df.dropna(subset=["speaker_clean"])
line_counts = labelled["speaker_clean"].value_counts()


def describe_framing(name, labels):
    """Classes, coverage and majority-class baseline of one candidate target definition."""
    counts = labels.value_counts()
    return {
        "framing": name,
        "n_classes": labels.nunique(),
        "rows_kept": len(labels),
        "pct_rows_kept": round(len(labels) / len(labelled) * 100, 1),
        "majority_class": counts.idxmax(),
        "baseline_accuracy_%": round(counts.max() / counts.sum() * 100, 1),
        "smallest_class_n": counts.min(),
    }


framings = [describe_framing("speaker_clean (all)", labelled["speaker_clean"])]
for k in (5, 10, 15):
    top_k = line_counts.head(k).index
    framings.append(describe_framing(f"top-{k} only (drop the rest)",
                                     labelled.loc[labelled["speaker_clean"].isin(top_k), "speaker_clean"]))
    framings.append(describe_framing(f"top-{k} + Other",
                                     labelled["speaker_clean"].where(labelled["speaker_clean"].isin(top_k), "Other")))
for min_lines in (20, 50):
    keep = line_counts[line_counts >= min_lines].index
    framings.append(describe_framing(f">= {min_lines} lines only",
                                     labelled.loc[labelled["speaker_clean"].isin(keep), "speaker_clean"]))

framing_table = pd.DataFrame(framings).set_index("framing")
framing_table
```

**Interpretation.** Each framing answers a slightly different question, and each has its own baseline:

- **Top-5 only** (the family) keeps about 61% of the lines with classes of 650+ lines each; it is the most
  learnable version and the closest to "recognize the main characters' voices", with a baseline of about 36%.
- **Top-k + Other** keeps every line but makes `"Other"` the majority class, so the baseline rises to about
  34-39% and part of the task becomes "main character vs. anyone else".
- **Top-10/15 only** or **≥ N lines only** add recurring side characters, but their classes are tiny (a few dozen
  lines, often from a single episode), which makes them hard to learn and noisy to evaluate.

No framing is chosen here. The recommendation for modeling is to compare at least *top-5 only*, *top-5 +
Other* and *top-10 + Other*, always reporting accuracy next to that framing's own baseline and with
imbalance-aware metrics (macro-F1, per-class recall), since accuracy alone rewards predicting the big classes.

### 9.2 `dialogue_clean` / `stage_direction_text`: separating narration from spoken text

**Definition.** `clean_dialogue()` turns each raw line into the spoken text plus the narration removed from it,
in four ordered steps:

1. **Layout:** remove raw HTML tags, drop the stray leading `:` and collapse newlines/indentation/NBSP into single spaces.
2. **Marked narration:** extract `[...]`, `(...)` and `*...*` spans, plus brackets left unclosed or unopened by
   the scrape (e.g. `"offscreen) Losing him."`).
3. **Unmarked narration (newline-led transcript style only):** in the episodes formatted with the action
   *before* the speech, extract the leading run of lowercase words that ends right before the first capitalized
   word (*"rubs his eyes. | What, Rick?"*), and lowercase action sentences that start after a sentence end.
4. **Tidy:** collapse spaces again and re-attach punctuation orphaned by the removals.

`stage_direction_text` keeps what was removed (for auditing), `has_stage_direction` flags rows where anything
was removed, and `is_empty_after_clean` flags rows with no spoken text left.

**Intuition.** A future speaker classifier should be trained on what the character actually *says*. Section 8
showed that narration density depends on the episode's transcription style, so leaving it in would inject a
spurious "formatting → speaker" signal.

```python
BRACKET_SPAN = re.compile(r"\[[^\[\]]*\]|\([^()]*\)|\*[^*]*\*")
UNOPENED_CLOSE = re.compile(r"^[^()\[\]]*[)\]]")      # "offscreen) Losing him."
UNCLOSED_OPEN = re.compile(r"[(\[*][^()\[\]*]*$")       # "... (Beth adjusts the organs again."
GLUED_WORDS = re.compile(r"(?<=[a-z])(?=[A-Z])")         # "speakerOh" -> "speaker Oh"
LEADING_ACTION = re.compile(r"^((?:[a-z][\w'’,\-]*[.!?,;:]?\s+)+?)(?=[A-Z\"“‘'¿¡])")
MID_ACTION = re.compile(r"(?<=[.!?])\s+((?:[a-z][\w'’,\-]*[.!?,;:]?\s+)+?)(?=[A-Z])")


def clean_dialogue(text: str) -> tuple[str, list[str]]:
    """Split a raw transcript line into (spoken_text, removed_narration_fragments)."""
    removed = []

    def _take(match):
        removed.append(match.group(match.lastindex or 0).strip(" []()*"))
        return " "

    transcript_style = text.startswith("\n")
    s = re.sub(HTML_TAG_RE, " ", text)
    s = re.sub(r"^\s*:\s*", "", s)
    s = re.sub(r"\s+", " ", s).strip()

    s = BRACKET_SPAN.sub(_take, s)
    s = UNOPENED_CLOSE.sub(_take, s)
    s = UNCLOSED_OPEN.sub(_take, s)

    if transcript_style:
        first, _, rest = s.partition(" ")
        s = (GLUED_WORDS.sub(" ", first) + " " + rest).strip()
        s = LEADING_ACTION.sub(_take, s, count=1)
        s = MID_ACTION.sub(_take, s)

    s = re.sub(r"\s+([,.!?;:])", r"\1", s)
    s = re.sub(r"[,;:]+([!?.])", r"\1", s)  # "Oh, [bleep]!" -> "Oh!"
    s = re.sub(r"^[,.;:\s]+|\s+", lambda m: "" if m.start() == 0 else " ", s).strip()
    return s, [frag for frag in removed if frag]


_cleaned = df["dialogue"].map(clean_dialogue)
df["dialogue_clean"] = _cleaned.str[0]
df["stage_direction_text"] = _cleaned.str[1].map(lambda frags: " | ".join(frags) if frags else np.nan)
df["has_stage_direction"] = df["stage_direction_text"].notna()
df["is_empty_after_clean"] = df["dialogue_clean"].eq("")

markers_clean = text_markers(df["dialogue_clean"])
before_after = pd.DataFrame({
    "pct_rows_raw": markers_raw.mean().mul(100),
    "pct_rows_clean": markers_clean.mean().mul(100),
}).round(2)
before_after.loc["lowercase sentence start after . ! ? (residual check)"] = [
    df["dialogue"].str.contains(r"[.!?][\s\xa0]+[a-z]").mean() * 100,
    df["dialogue_clean"].str.contains(r"[.!?][\s\xa0]+[a-z]").mean() * 100,
]

print(f"Rows where narration was removed: {df['has_stage_direction'].sum():,} ({df['has_stage_direction'].mean()*100:.1f}%)")
print(f"Rows with no spoken text left (pure narration or scrape junk): {df['is_empty_after_clean'].sum():,}")
before_after.round(2)
```

```python
examples = df.loc[df["has_stage_direction"], ["dialogue", "dialogue_clean", "stage_direction_text"]].sample(
    8, random_state=RANDOM_SEED
)
for _, row in examples.iterrows():
    print("RAW:    ", repr(row["dialogue"][:110]))
    print("SPOKEN: ", repr(row["dialogue_clean"][:110]))
    print("REMOVED:", repr(row["stage_direction_text"][:110]))
    print()

print("Residual examples (lowercase sentence starts still present after cleaning):")
for txt in df.loc[df["dialogue_clean"].str.contains(r"[.!?][\s\xa0]+[a-z]"), "dialogue_clean"].sample(
    5, random_state=RANDOM_SEED
):
    print("-", repr(txt[:120]))
```

**Evidence.** Layout artifacts (leading `:`, newlines, padding) disappear completely, and bracketed/
parenthesized narration drops to a handful of rows. On the random sample above the spoken text and the
removed narration are separated correctly in most cases. Rows that end up empty are lines that contained *only*
narration or sound cues (`[Gurgling]`) or scrape junk (`"["`, blank strings); they carry no spoken text, so they
are the natural candidates to drop before vectorizing (a modeling decision, not applied here).

**Limitations.** This is a large improvement on the original single regex (which caught only the first
sentence of about 1.5% of rows), but it is still a heuristic:
- **Unmarked narration** written in normal sentence case outside the newline-led episodes cannot be told apart
  from speech without a language model. The residual examples above show both kinds: some are leftover actions
  (*"Yeah. puts an arm around Beth and Jacob I guess..."*) and others are real speech (*"That... is my groin's
  user."*). Applying the lowercase-action rule to every row would also cut genuine lines such as *"Help me! help
  me please!"*, so it is restricted to the newline-led episodes where the pattern is reliable.
- **Censor cues** such as `[bleep]` are removed as non-speech, which also removes the information that a word
  was censored there.
- In transcript-style rows an action mentioning a name (*"spills alcohol on Morty's bed Come on"*) is cut at
  the first capitalized word, leaving part of the action in the spoken text.

So narration is now *substantially reduced*, not *eliminated*. Measuring how much the residual hurts (e.g.
comparing models trained on `dialogue` vs `dialogue_clean`) is part of the modeling stage.

### 9.3 Rejected candidate: `turn_fraction` (and any episode-derived feature)

An earlier version of this notebook proposed `turn_fraction` = (position of the line inside its episode) /
(total number of lines in the episode) as a feature. It is **excluded**, because it violates the
prediction-time information set defined in Section 1. Computing it requires knowing:

1. which episode the line belongs to;
2. where the line sits inside that episode;
3. how many lines the **whole** episode ends up having, which is information from *after* the line was spoken.

None of that exists when the only input is a dialogue line, so a model using it would look better in validation
than it could ever be in use (leakage). The same applies to `episode no.` itself and to any statistic computed
from other lines of the conversation or episode (neighbouring speakers, per-episode speaker counts, etc.).

The episode information is still useful **outside** the feature set: it describes the data (Sections 3.1 and 8),
and it is the natural grouping key for a validation split that does not let episode-specific vocabulary leak
between training and evaluation (Section 11). A model that does receive episode or conversational context is a
legitimate but **different** problem (a separate "contextual" experiment) and must be framed and reported as
such.

### 9.4 Lexical ratios: `question_ratio`, `exclamation_ratio`

**Definition.** For each row, the count of `?` and `!` characters in `dialogue_clean` divided by the number of
words in `dialogue_clean` (`clean_word_count`; rows with zero words are excluded to avoid division by zero).

**Intuition.** Punctuation style is often part of a character's "voice", for example a character written as
anxious/questioning versus one written as brash/exclamatory, and is cheap to compute from the text alone. It is computed on
`dialogue_clean` so that punctuation inside stage directions (e.g. `(Gasp!)`) is not credited to the speaker.


```python
df["clean_word_count"] = df["dialogue_clean"].str.split().str.len()
has_words = df["clean_word_count"] > 0
df["question_ratio"] = np.where(has_words, df["dialogue_clean"].str.count(r"\?") / df["clean_word_count"], np.nan)
df["exclamation_ratio"] = np.where(has_words, df["dialogue_clean"].str.count(r"!") / df["clean_word_count"], np.nan)

ratio_by_group = (
    df[df["speaker_group"].notna() & has_words]
    .groupby("speaker_group")[["question_ratio", "exclamation_ratio"]]
    .mean()
    .loc[lambda d: d.index != "Other"]
    .sort_values("question_ratio", ascending=False)
)

fig, ax = plt.subplots(figsize=(10, 5.5))
ratio_by_group.plot(kind="barh", ax=ax)
ax.set_title("Average Question-Mark and Exclamation-Mark Ratio by Speaker (spoken text only)")
ax.set_xlabel("Marks per word")
ax.set_ylabel("Speaker")
plt.tight_layout()
plt.show()
```

**Evidence.** The average ratios differ between speakers in the chart above, a per-character stylistic
difference that raw word counts (Section 8) did not clearly show. These are **averages**: individual lines
overlap heavily, and several side characters contribute only about 40 lines, so their bars are noisy. The
differences are descriptive and need to be confirmed on held-out data.

**Limitations.** Ratios computed on very short lines are noisy (one `?` on a 2-word line produces a large
ratio from very little evidence); punctuation from the original scrape is not guaranteed to be perfectly
transcribed. This should be paired with a minimum word-count filter if used as a model feature.

### 9.5 Character-name and address-term mentions: voice vs. conversational context

**Definition.** Two families of tokens inside the spoken text (`dialogue_clean`):

- **Character names**, built from the data rather than typed by hand: every capitalized token (3+ letters) in a
  `speaker_clean` label with at least 2 lines, kept only if it appears in the dialogue at least twice and is
  written in lowercase at most 15% of the time. The second condition separates names (*Morty*, *Jessica*,
  *Unity*) from ordinary words that also occur in labels (*Alien*, *Toxic*, *Pickle*, *Man*).
- **Address terms** (*grandpa, dad, mom, sweetie, honey, son*, …): kinship/affection words used to address
  someone, which encode *who is being spoken to* just like names do.

From these, `mentions_name` and `mentions_address_term` flag each row, and `mask_names()` replaces character
names with a `NAME` placeholder to produce `dialogue_masked`.

**Intuition.** *"What, Rick?"* is very likely Morty, but not because of how Morty talks: it is because Morty
talks **to** Rick. Names and address terms can be predictive while capturing **conversational structure (who
talks to whom)** rather than the character's **own voice**. The project premise is about voice, so this signal
has to be measured explicitly before it silently dominates a bag-of-words model.

```python
speaker_line_counts = df["speaker_clean"].value_counts()
label_tokens = {
    tok
    for label in speaker_line_counts[speaker_line_counts >= 2].index
    for tok in re.findall(r"\b[A-Z][a-z]{2,}\b", label)
} - {"Mrs"}

token_counts = pd.Series(re.findall(r"[A-Za-z]+", " ".join(df["dialogue_clean"]))).value_counts()
name_stats = pd.DataFrame({
    "capitalized": [token_counts.get(t, 0) for t in sorted(label_tokens)],
    "lowercase": [token_counts.get(t.lower(), 0) for t in sorted(label_tokens)],
}, index=sorted(label_tokens))
name_stats["lowercase_share"] = name_stats["lowercase"] / (name_stats["capitalized"] + name_stats["lowercase"])
CHARACTER_NAMES = sorted(
    name_stats.query("capitalized >= 2 and lowercase_share <= 0.15").index, key=len, reverse=True
)

NAME_RE = re.compile(r"\b(?:" + "|".join(CHARACTER_NAMES) + r")\b")
ADDRESS_TERMS = ["grandpa", "grandson", "dad", "daddy", "mom", "mommy", "sweetie", "honey", "son", "sis", "kids"]
ADDRESS_RE = re.compile(r"\b(?:" + "|".join(ADDRESS_TERMS) + r")\b", flags=re.IGNORECASE)


def mask_names(text: str) -> str:
    """Replace character names with a neutral placeholder so a model cannot use who is being addressed."""
    return NAME_RE.sub("NAME", text)


df["mentions_name"] = df["dialogue_clean"].str.contains(NAME_RE)
df["mentions_address_term"] = df["dialogue_clean"].str.contains(ADDRESS_RE)
df["dialogue_masked"] = df["dialogue_clean"].map(mask_names)

print(f"Character-name lexicon: {len(CHARACTER_NAMES)} names, e.g. {sorted(CHARACTER_NAMES, key=lambda t: -name_stats.loc[t, 'capitalized'])[:12]}")
print(f"Lines mentioning a character name: {df['mentions_name'].mean()*100:.1f}%")
print(f"Lines using an address term:       {df['mentions_address_term'].mean()*100:.1f}%")
print(f"Lines with either:                 {(df['mentions_name'] | df['mentions_address_term']).mean()*100:.1f}%")
print()
for raw_text in df.loc[df["mentions_name"], "dialogue_clean"].head(3):
    print("ORIGINAL:", raw_text[:90])
    print("MASKED:  ", mask_names(raw_text)[:90])
```

```python
MAIN_CAST = ["Rick", "Morty", "Jerry", "Beth", "Summer"]
mention_cols = {name: rf"\b{name}\b" for name in MAIN_CAST}
mention_cols.update({"grandpa": r"(?i)\bgrandpa\b", "dad/daddy": r"(?i)\bdad(?:dy)?\b", "mom/mommy": r"(?i)\bmom(?:my)?\b"})

main_cast_lines = df[df["speaker_clean"].isin(MAIN_CAST)]
mention_matrix = pd.DataFrame({
    col: main_cast_lines["dialogue_clean"].str.contains(pattern).groupby(main_cast_lines["speaker_clean"]).mean() * 100
    for col, pattern in mention_cols.items()
}).loc[MAIN_CAST]
row_sizes = main_cast_lines["speaker_clean"].value_counts()

fig, ax = plt.subplots(figsize=(10, 4.8))
sns.heatmap(mention_matrix, annot=True, fmt=".1f", cmap="Blues", cbar_kws={"label": "% of the speaker's lines"}, ax=ax)
ax.set_title("Who Mentions Whom: % of Each Speaker's Lines Containing a Name or Address Term")
ax.set_xlabel("Token mentioned in the line")
ax.set_ylabel("Speaker (n lines)")
ax.set_yticks(np.arange(len(MAIN_CAST)) + 0.5)
ax.set_yticklabels([f"{name} (n={row_sizes[name]})" for name in MAIN_CAST], rotation=0)
plt.tight_layout()
plt.show()

self_mention = pd.Series({name: mention_matrix.loc[name, name] for name in MAIN_CAST}).round(1)
print("Self-mention rate (% of a speaker's lines containing their own name):")
print(self_mention.to_string())
```

**Evidence.** Mentions are frequent and highly structured:

- About **29%** of all lines mention a character name and about **6%** use an address term (one line in three
  has at least one of them).
- The heatmap is close to a map of *relationships*, not of *styles*: **30%** of Rick's lines contain "Morty"
  and **23.5%** of Morty's lines contain "Rick", while characters rarely say their own name (self-mention
  1-3%).
- Address terms encode family roles: Summer says *grandpa* (8.6%) and *dad* (8.0%); Beth says *dad* (10.4%,
  to Rick) and *Jerry* (11.5%); Morty never names Jerry or Beth (0.0%) but says *dad*/*mom* instead.

**Interpretation and limitations.** A bag-of-words model will find these tokens extremely useful (e.g. "Morty"
→ predict Rick), but that is **association through conversational structure**: it tells us who the speaker
usually addresses, not how they talk. It would also break on text where the usual partner is absent. This is
why `dialogue_masked` is kept as a second input variant: comparing models trained on `dialogue_clean` vs.
`dialogue_masked` during modeling will measure how much of the accuracy comes from **voice** and how much from
**context**. Limitations: the name lexicon is built from speaker labels, so characters who are mentioned but
never speak in this scrape (and nicknames) are not masked, and address terms are left unmasked here because
they are also ordinary words (*son*, *honey*); masking them is a variant to test, not a decision made here.

### 9.6 Distinctive vocabulary per top speaker: words, two-word expressions, and the effect of masking names

**Definition.** For each of the top-5 speakers, unigrams **and bigrams** (`ngram_range=(1, 2)`) from the spoken
text are ranked with the **log-odds ratio with an informative Dirichlet prior** (Monroe, Colaresi & Quinn,
2008, *"Fightin' Words"*). For a term *w*, speaker *i* is compared against all other top-5 speakers *j*, using
the term's frequency in the whole corpus as the prior $\alpha_w$:

$$\delta_w = \log\frac{y_{iw}+\alpha_w}{n_i+\alpha_0-y_{iw}-\alpha_w} - \log\frac{y_{jw}+\alpha_w}{n_j+\alpha_0-y_{jw}-\alpha_w},
\qquad z_w = \frac{\delta_w}{\sqrt{\frac{1}{y_{iw}+\alpha_w}+\frac{1}{y_{jw}+\alpha_w}}}$$

A large positive $z_w$ means the term is used by speaker *i* more than chance would explain, given how common
it is overall. Unlike a raw frequency ratio, this does not over-reward rare words (a word said 3 times by one
character) nor very common ones. The ranking is computed twice: on `dialogue_clean` and on `dialogue_masked`
(names replaced by `NAME`, Section 9.5). Stop words are **kept** on purpose: function words, fillers and
stutters are part of a character's voice.

**Intuition.** Catchphrases and recurring expressions (*"aw geez"*) are often two words long, so unigrams alone
miss them. Comparing original vs. masked rankings shows which distinctive terms are *voice* and which are only
*who the character talks to*.

```python
def fightin_words(texts: pd.Series, groups: pd.Series, ngram_range=(1, 2), min_df=5, top_k=10):
    """Per-group z-scores of the log-odds ratio with an informative Dirichlet prior (Monroe et al., 2008)."""
    vec = CountVectorizer(ngram_range=ngram_range, min_df=min_df, lowercase=True)
    counts = vec.fit_transform(texts)
    terms = vec.get_feature_names_out()
    prior = np.asarray(counts.sum(axis=0)).ravel().astype(float)  # alpha_w: corpus frequency of each term
    alpha0 = prior.sum()
    z_by_group, top_terms = {}, {}
    for group in groups.unique():
        mask = (groups == group).values
        y_i = np.asarray(counts[mask].sum(axis=0)).ravel()
        y_j = np.asarray(counts[~mask].sum(axis=0)).ravel()
        n_i, n_j = y_i.sum(), y_j.sum()
        delta = (np.log((y_i + prior) / (n_i + alpha0 - y_i - prior))
                 - np.log((y_j + prior) / (n_j + alpha0 - y_j - prior)))
        z = delta / np.sqrt(1 / (y_i + prior) + 1 / (y_j + prior))
        z_by_group[group] = pd.Series(z, index=terms)
        top_terms[group] = list(z_by_group[group].nlargest(top_k).index)
    return pd.DataFrame(z_by_group), pd.DataFrame(top_terms)


corpus_top5 = df[df["speaker_clean"].isin(top5_names) & ~df["is_empty_after_clean"]]
z_original, top_original = fightin_words(corpus_top5["dialogue_clean"], corpus_top5["speaker_clean"])
z_masked, top_masked = fightin_words(corpus_top5["dialogue_masked"], corpus_top5["speaker_clean"])

print("Most distinctive unigrams/bigrams per speaker, ORIGINAL spoken text:")
display(top_original[list(top5_names)])
print("Most distinctive unigrams/bigrams per speaker, NAMES MASKED:")
display(top_masked[list(top5_names)])
```

**Evidence.** The rankings separate into three kinds of signal, and only one of them is clearly *voice*:

- **Conversational context (names and address terms).** On the original text Rick's list is dominated by
  *morty* and bigrams around it (*morty you*, *right morty*, *on morty*), Morty's by *rick*, Beth's by *jerry*
  and *dad*, Summer's by *grandpa* / *grandpa rick*. These are the who-talks-to-whom associations from
  Section 9.5. After masking, Rick's top term becomes the placeholder *name* itself: Rick addresses people by
  name far more than the others, which is a habit of his *speech* even though *which* name it is was context.
- **Voice / catchphrases.** Some terms survive masking and look like genuine style: Morty's *geez*, *oh geez*,
  *oh man*, *you know*, *ow*, *ohh*; Rick's *shit*; Summer's *oh my*, *my god*; Beth's *love you*. Bigrams
  matter here: *oh geez* and *oh man* are more characteristic than *oh* alone.
- **Episode-specific plot vocabulary.** Jerry's *apples* and *pluto*, Summer's *needful* and Beth's *tommy*
  each occur in a single episode, and Jerry's *christmas* mostly in one (7 of 9 lines). These are memorization
  candidates: a model would profit from them only if lines from the same episode appear in both training and
  evaluation, which is a concrete argument for **episode-grouped validation** (Section 11). Jerry's *starts
  beeping* is a different problem: it is residual narration (*"The heart rate ... starts beeping"*) written in
  sentence case, which the Section 9.2 heuristic cannot detect.

The table below quantifies the masking effect: every speaker loses statistically distinctive terms once names
are masked (Morty the most, 46 → 31), and for Rick half of his top-20 terms contain a name.

**Interpretation and limitations.** These are **descriptive, in-sample associations**: they show that the
speakers differ in vocabulary on this data, i.e. *candidate* lexical signals. Whether they help predict the
speaker of **unseen** lines must be checked on validation data that was not used to build the vocabulary; any
vectorizer or prior must then be fitted on the training split only. Other caveats: the default tokenizer splits
contractions (`I'll` → *ll*), z-scores assume independent tokens (lines from the same scene are not), and
`dialogue_clean` still holds some residual narration (Section 9.2).

```python
name_terms = {t.lower() for t in CHARACTER_NAMES}
top20_original = {s: z_original[s].nlargest(20).index for s in top5_names}
masking_effect = pd.DataFrame({
    "significant_terms_original (z>1.96)": (z_original[list(top5_names)] > 1.96).sum(),
    "significant_terms_masked (z>1.96)": (z_masked[list(top5_names)] > 1.96).sum(),
    "top20_terms_containing_a_name": pd.Series({
        s: sum(any(tok in name_terms for tok in term.split()) for term in terms) for s, terms in top20_original.items()
    }),
}).loc[list(top5_names)]
masking_effect
```

### 9.7 Fillers, interjections and lexical diversity per speaker

**Definition.** (a) The rate of common fillers/interjections per **1,000 words** of each speaker's spoken text,
so speakers with more lines are not favoured by raw counts. (b) **Lexical diversity** as the type-token ratio
(TTR = distinct words / total words) computed on **random samples of exactly 1,500 tokens** per speaker (30
repetitions, fixed seed). TTR falls as text gets longer, so comparing raw TTR between Rick (thousands of lines)
and Summer (hundreds) would mostly measure corpus size; fixing the sample size removes that bias.

**Intuition.** Hesitation and stuttering (*uh, um, geez*) versus assertive or exclamatory openers (*oh, hey,
listen*), and a richer versus more repetitive vocabulary, are stylistic traits that do not depend on who is
being addressed, i.e. candidate *voice* signals.

```python
INTERJECTIONS = ["uh", "um", "oh", "aw", "geez", "ooh", "whoa", "wow", "hey", "huh", "yeah", "okay", "listen"]
tokenize = CountVectorizer(lowercase=True, token_pattern=r"(?u)\b[a-z']+\b").build_analyzer()
tokens_by_speaker = {s: [tok for line in corpus_top5.loc[corpus_top5["speaker_clean"] == s, "dialogue_clean"]
                         for tok in tokenize(line)] for s in top5_names}

interjection_rate = pd.DataFrame({
    s: {w: toks.count(w) / len(toks) * 1000 for w in INTERJECTIONS} for s, toks in tokens_by_speaker.items()
}).T

fig, ax = plt.subplots(figsize=(11, 4.2))
sns.heatmap(interjection_rate, annot=True, fmt=".1f", cmap="Purples", cbar_kws={"label": "occurrences per 1,000 words"}, ax=ax)
ax.set_title("Fillers and Interjections per 1,000 Spoken Words (Top-5 Speakers)")
ax.set_xlabel("Token")
ax.set_ylabel("Speaker (total words)")
ax.set_yticks(np.arange(len(top5_names)) + 0.5)
ax.set_yticklabels([f"{s} ({len(tokens_by_speaker[s]):,})" for s in top5_names], rotation=0)
plt.tight_layout()
plt.show()

rng = np.random.default_rng(RANDOM_SEED)
SAMPLE_TOKENS, N_REPEATS = 1500, 30
ttr = pd.DataFrame({
    s: [len(set(rng.choice(toks, SAMPLE_TOKENS, replace=False))) / SAMPLE_TOKENS for _ in range(N_REPEATS)]
    for s, toks in tokens_by_speaker.items()
})
ttr_summary = ttr.agg(["mean", "std"]).T.round(3).assign(total_tokens={s: len(t) for s, t in tokens_by_speaker.items()})
ttr_summary.sort_values("mean", ascending=False)
```

**Evidence.** Normalized per 1,000 words, the fillers show clear stylistic profiles that do not depend on
names:

- **Morty** is the hesitant speaker: the highest rates of *oh* (11.5), *um* (2.1), *aw* (1.1) and especially
  *geez* (2.7, versus 0.3 or less for everyone else).
- **Jerry** has the highest *uh* (4.8); **Beth** the highest *okay* (6.6); **Summer** the highest *yeah* (6.4).
- **Rick** almost never says *um* (0.2) or *aw* (0.1) and has the highest *listen* (1.5).

Lexical diversity on equal-size samples ranks **Rick** highest (TTR ≈ 0.41), **Jerry** and **Summer** in the
middle (≈ 0.38) and **Morty** and **Beth** lowest (≈ 0.35-0.36); the standard deviations across the 30 samples
(≈ 0.01) are small compared with the gap between Rick and Morty/Beth.

**Limitations.** These are averages over many lines: an individual short line usually contains none of these
tokens, so they are weak signals per line and useful mainly in combination with the text representation. TTR
on pooled tokens says nothing about a single line. Profiles are only reliable for speakers with thousands of
words (the top-5); they were not computed for minor characters.

## 11. Conclusions, Limitations, and Roadmap

**Problem framing (fixed).** Given **only the text of one dialogue line**, predict the character who said it:
a multiclass text-classification problem. `episode no.` and any position or conversation-derived variable are
not available at prediction time; they are used here only to describe the data and to design validation.

**Main findings.**
- The dataset is a partial transcript scrape: 9,618 rows across 39 of 51 episode numbers (12 missing) with very
  uneven rows per episode.
- The target is extremely high-cardinality: 976 raw labels, **840** after normalization, with 44% of speakers
  having a single line. The majority-class baseline for `speaker_clean` is **21.7%** (Rick).
- Consolidated framings change the baseline: *top-5 only* keeps 61% of rows with a 35.7% baseline; *top-k +
  Other* keeps every row but makes `"Other"` the majority class (33.6%-39.2% baseline).
- The `dialogue` field is not pure speech: about 93% of rows carry layout artifacts (stray `:`, newlines, HTML
  tags) and about 14% contain narration. The cleaning step removes all layout artifacts and nearly all marked
  narration; about 2.5% of rows still show signs of unmarked narration.
- Narration density depends on the **episode's transcription style** (0%-72% per episode), so it is a
  formatting signal correlated with speakers through episodes, not part of anyone's voice.
- About **29%** of lines name a character and about **6%** use an address term. These map relationships
  (30% of Rick's lines say "Morty", 23.5% of Morty's say "Rick"), i.e. conversational context rather than voice.
- Once names are masked, candidate voice signals remain: fillers and catchphrases (Morty's *geez*, *oh geez*,
  *oh man*; Rick's low use of *um*/*aw*), and differences in lexical diversity (Rick highest). Some distinctive
  terms are episode-specific plot words (*pluto*, *apples*, *needful*). All of these are **descriptive in-sample
  associations**, not evidence of predictive power.
- Line length and punctuation ratios differ only modestly between speakers: weak signals on their own.

**Quality problems that must be treated before modeling.**
- Use `dialogue_clean` (or `dialogue_masked`) instead of raw `dialogue`; drop the 174 rows with no spoken text
  left (`is_empty_after_clean`) and the 3 rows without a speaker.
- Residual unmarked narration (Section 9.2) remains; measure its effect rather than assume it is negligible.
- Decide whether exact duplicate lines (394) are kept as genuine repeats; they must not be split across
  training and evaluation.
- Some `speaker` labels are not characters (e.g. scene headers such as *After-Credit Scene*) or are numbered
  variants (*Summer 1*, *Morty 2*); review them if minor classes are kept.

**Variables and relationships worth carrying into modeling** (see the inventory in Section 10).
- Text representations (word and two-word counts / TF-IDF) built from `dialogue_clean`, plus the weak numeric
  features (`clean_word_count`, `question_ratio`, `exclamation_ratio`).
- `dialogue_masked` as an ablation, to separate voice from conversational context.
- **Excluded:** `episode no.`, `turn_fraction` and any episode/position feature, and `has_stage_direction`
  (transcription-style signal).

**Limitations, biases, and open questions.**
- Only 39 of 51 numbered episodes, unevenly covered: conclusions describe this scrape, not the whole show.
- Speaker labels come from fan transcripts and were not verified; mislabeling cannot be ruled out.
- Minor characters have a few dozen lines at most, often from one episode; they cannot be learned reliably.
- The lexical analyses fit their vocabularies on the full dataset for description only; they say nothing about
  generalization.

**Hypotheses to validate during modeling (not decided here).**
1. **Target framing:** compare *top-5 only*, *top-5 + Other* and *top-10 + Other*, each against its own
   majority-class baseline, using macro-F1 and per-class recall in addition to accuracy.
2. **Validation design:** use episode-grouped cross-validation (e.g. `StratifiedGroupKFold` with `episode no.`
   as the group) so that plot-specific vocabulary and duplicate lines from one episode cannot appear in both
   training and evaluation; compare with a random stratified split to measure how optimistic the latter is.
3. **Leakage prevention:** fit every vectorizer, IDF weighting, scaler, encoder or feature selection on the
   training folds only (inside a `Pipeline`).
4. **Voice vs. context:** compare models trained on `dialogue_clean` vs. `dialogue_masked`; the gap estimates
   how much performance comes from names rather than voice.
5. **Cleaning:** compare raw `dialogue` vs. `dialogue_clean` to measure the effect of narration and layout noise.
6. **Unigrams vs. unigrams + bigrams:** test whether two-word expressions add predictive value.
7. **Contextual experiment (optional, separate framing):** a model that also receives the episode or previous
   lines would answer a different question and must be reported separately from the text-only model.
